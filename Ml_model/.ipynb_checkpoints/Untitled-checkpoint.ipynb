```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, 
                             f1_score, roc_auc_score, confusion_matrix, 
                             classification_report, roc_curve)
from sklearn.utils.class_weight import compute_class_weight
import xgboost as xgb
import warnings
import gc

warnings.filterwarnings('ignore')

# Set visualization style
sns.set_style('whitegrid')
plt.rcParams['figure.figsize'] = (12, 8)

print("Libraries imported successfully!")
```

```python
dtype_dict = {
    'FLOW_START_MILLISECONDS': 'int64',
    'FLOW_END_MILLISECONDS': 'int64',
    'IPV4_SRC_ADDR': 'category',
    'L4_SRC_PORT': 'uint16',
    'IPV4_DST_ADDR': 'category',
    'L4_DST_PORT': 'uint16',
    'PROTOCOL': 'uint8',
    'L7_PROTO': 'float32',
    'IN_BYTES': 'uint32',
    'IN_PKTS': 'uint32',
    'OUT_BYTES': 'uint32',
    'OUT_PKTS': 'uint32',
    'TCP_FLAGS': 'uint8',
    'CLIENT_TCP_FLAGS': 'uint8',
    'SERVER_TCP_FLAGS': 'uint8',
    'FLOW_DURATION_MILLISECONDS': 'int32',
    'DURATION_IN': 'int32',
    'DURATION_OUT': 'int32',
    'MIN_TTL': 'uint8',
    'MAX_TTL': 'uint8',
    'LONGEST_FLOW_PKT': 'uint16',
    'SHORTEST_FLOW_PKT': 'uint16',
    'MIN_IP_PKT_LEN': 'uint16',
    'MAX_IP_PKT_LEN': 'uint16',
    'SRC_TO_DST_SECOND_BYTES': 'float32',
    'DST_TO_SRC_SECOND_BYTES': 'float32',
    'RETRANSMITTED_IN_BYTES': 'uint32',
    'RETRANSMITTED_IN_PKTS': 'uint16',
    'RETRANSMITTED_OUT_BYTES': 'uint32',
    'RETRANSMITTED_OUT_PKTS': 'uint16',
    'SRC_TO_DST_AVG_THROUGHPUT': 'float32',
    'DST_TO_SRC_AVG_THROUGHPUT': 'float32',
    'NUM_PKTS_UP_TO_128_BYTES': 'uint16',
    'NUM_PKTS_128_TO_256_BYTES': 'uint16',
    'NUM_PKTS_256_TO_512_BYTES': 'uint16',
    'NUM_PKTS_512_TO_1024_BYTES': 'uint16',
    'NUM_PKTS_1024_TO_1514_BYTES': 'uint16',
    'TCP_WIN_MAX_IN': 'uint32',
    'TCP_WIN_MAX_OUT': 'uint32',
    'ICMP_TYPE': 'uint8',
    'ICMP_IPV4_TYPE': 'uint8',
    'DNS_QUERY_ID': 'uint16',
    'DNS_QUERY_TYPE': 'uint16',
    'DNS_TTL_ANSWER': 'uint32',
    'FTP_COMMAND_RET_CODE': 'uint16',
    'SRC_TO_DST_IAT_MIN': 'float32',
    'SRC_TO_DST_IAT_MAX': 'float32',
    'SRC_TO_DST_IAT_AVG': 'float32',
    'SRC_TO_DST_IAT_STDDEV': 'float32',
    'DST_TO_SRC_IAT_MIN': 'float32',
    'DST_TO_SRC_IAT_MAX': 'float32',
    'DST_TO_SRC_IAT_AVG': 'float32',
    'DST_TO_SRC_IAT_STDDEV': 'float32',
    'Label': 'uint8',
    'Attack': 'category'
}

print("Loading dataset with optimized data types...")
print("This may take a moment for large files...\n")

# Load with optimized dtypes
df = pd.read_csv('data/NF-UNSW-NB15-v3.csv', dtype=dtype_dict, low_memory=False)

print(f"✓ Dataset loaded successfully!")
print(f"Dataset shape: {df.shape}")
print(f"Memory usage: {df.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
```

```python
print("First few rows:")
print(df.head(3))
print(f"\nColumn names ({len(df.columns)} total):")
print(df.columns.tolist())
print(f"\nData types:")
print(df.dtypes)
print(f"\nTarget variable distribution:")
print(df['Label'].value_counts())
print(f"\nAttack type distribution:")
print(df['Attack'].value_counts())
```

```python
print("Checking for infinite and missing values...")

# Identify categorical and numerical columns
categorical_cols = df.select_dtypes(include=['category']).columns.tolist()
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()

print(f"Categorical columns: {categorical_cols}")
print(f"Numerical columns: {len(numerical_cols)} total")

# Replace infinite values with NaN (only in numerical columns)
for col in numerical_cols:
    df[col] = df[col].replace([np.inf, -np.inf], np.nan)

# Check for missing values
missing_counts = df.isnull().sum()
print(f"\nColumns with missing values:")
print(missing_counts[missing_counts > 0])

# Fill NaN values - handle categorical and numerical separately
for col in numerical_cols:
    if df[col].isnull().any():
        df[col].fillna(0, inplace=True)

# For categorical columns, fill with most frequent value or 'Unknown'
for col in categorical_cols:
    if df[col].isnull().any():
        # Add 'Unknown' category if needed
        if 'Unknown' not in df[col].cat.categories:
            df[col] = df[col].cat.add_categories(['Unknown'])
        df[col].fillna('Unknown', inplace=True)

print(f"\n✓ Data cleaning completed!")
print(f"Total NaN after filling: {df.isnull().sum().sum()}")
```

```python
print("Encoding categorical features...")

# IP addresses are already category type, get their codes
df['IPV4_SRC_ADDR_ENCODED'] = df['IPV4_SRC_ADDR'].cat.codes
df['IPV4_DST_ADDR_ENCODED'] = df['IPV4_DST_ADDR'].cat.codes

# Encode Attack type for multi-class classification
df['Attack_ENCODED'] = df['Attack'].cat.codes

# Store mapping for interpretation
attack_mapping = dict(enumerate(df['Attack'].cat.categories))
print(f"\n✓ Encoding completed!")
print(f"\nAttack type mapping ({len(attack_mapping)} classes):")
for code, attack in list(attack_mapping.items())[:10]:
    print(f"  {code}: {attack}")
if len(attack_mapping) > 10:
    print(f"  ... and {len(attack_mapping) - 10} more")
```

```python
print("="*70)
print("FEATURE PREPARATION WITH LEAKAGE PREVENTION")
print("="*70)

# CRITICAL: Remove features that cause data leakage
leakage_features = [
    'FLOW_START_MILLISECONDS',  # Timestamp - leaks attack timing
    'FLOW_END_MILLISECONDS',    # Timestamp - leaks attack timing
    'IPV4_SRC_ADDR',           # Categorical - already encoded
    'IPV4_DST_ADDR',           # Categorical - already encoded
    'Attack',                   # Target variable
    'Label',                    # Target variable
    'Attack_ENCODED'            # Encoded target
]

# Get initial feature columns
initial_features = [col for col in df.columns if col not in leakage_features]
print(f"\nInitial features after removing obvious leakage: {len(initial_features)}")

# Separate targets FIRST (before any analysis)
y_binary = df['Label'].values
y_multiclass = df['Attack_ENCODED'].values

# Create feature matrix with proper type conversion
X_temp = df[initial_features].copy()

# Convert categorical to numeric for leakage checking
for col in X_temp.select_dtypes(include=['category']).columns:
    X_temp[col] = X_temp[col].cat.codes

X_temp = X_temp.astype('float32')

# AUTOMATIC LEAKAGE DETECTION
print("\n🔍 Checking for data leakage...")
from scipy.stats import pointbiserialr

leaky_features = []
correlations = []

for col in X_temp.columns:
    try:
        # Remove any infinite values for correlation calculation
        valid_mask = np.isfinite(X_temp[col])
        if valid_mask.sum() > 0:
            corr, p_value = pointbiserialr(y_binary[valid_mask], X_temp.loc[valid_mask, col])
            abs_corr = abs(corr)
            correlations.append({
                'feature': col,
                'correlation': abs_corr,
                'p_value': p_value
            })
            
            # Flag features with suspiciously high correlation
            if abs_corr > 0.9:
                leaky_features.append(col)
    except Exception as e:
        pass

corr_df = pd.DataFrame(correlations).sort_values('correlation', ascending=False)

# Display results
print(f"\n📊 Top 15 features by correlation with Label:")
print(corr_df.head(15)[['feature', 'correlation']].to_string(index=False))

if len(leaky_features) > 0:
    print(f"\n⚠️  CRITICAL: {len(leaky_features)} features have correlation > 0.9 (LEAKAGE DETECTED)!")
    print("   Removing these features automatically:")
    for feat in leaky_features:
        corr_val = corr_df[corr_df['feature'] == feat]['correlation'].values[0]
        print(f"   - {feat} (corr: {corr_val:.4f})")
    
    # Remove leaky features
    final_features = [f for f in initial_features if f not in leaky_features]
else:
    print(f"\n✓ No obvious leakage detected (all correlations < 0.9)")
    final_features = initial_features

# Create final feature matrix
X = df[final_features].copy()

# Convert categorical to numeric
for col in X.select_dtypes(include=['category']).columns:
    X[col] = X[col].cat.codes

X = X.astype('float32')

print(f"\n{'='*70}")
print(f"✓ Feature preparation completed!")
print(f"{'='*70}")
print(f"Final feature count: {len(final_features)}")
print(f"Features removed: {len(initial_features) - len(final_features)}")
print(f"Feature matrix shape: {X.shape}")
print(f"Memory usage: {X.memory_usage(deep=True).sum() / 1024**2:.2f} MB")
print(f"{'='*70}")

# Store for later use
feature_cols = final_features

# Separate features and targets
X = df[feature_cols].astype('float32')  # Convert all features to float32 to save memory
y_binary = df['Label'].values
y_multiclass = df['Attack_ENCODED'].values

print(f"\n✓ Feature preparation completed!")
print(f"Feature matrix shape: {X.shape}")
print(f"Memory usage: {X.memory_usage(deep=True).sum() / 1024**2:.2f} MB")

# Clean up to free memory
del df
gc.collect()
```

```python
print("="*70)
print("ADVANCED DATA LEAKAGE DETECTION")
print("="*70)

# Calculate point-biserial correlation for binary target
from scipy.stats import pointbiserialr

print("\nAnalyzing feature correlations with target...")
correlations = []
for col in feature_cols:
    try:
        corr, p_value = pointbiserialr(y_binary, X[col])
        correlations.append({
            'feature': col,
            'correlation': abs(corr),
            'p_value': p_value
        })
    except Exception as e:
        pass

corr_df = pd.DataFrame(correlations).sort_values('correlation', ascending=False)

print("\n🔍 Features with SUSPICIOUS high correlation (potential leakage):")
suspicious = corr_df[corr_df['correlation'] > 0.9]
if len(suspicious) > 0:
    print(suspicious.head(20).to_string(index=False))
    print(f"\n⚠️  CRITICAL: {len(suspicious)} features have >0.9 correlation with target!")
    print("   These features likely cause data leakage and should be removed.")
    
    # Ask for confirmation to remove
    print("\n   Recommended action: Remove these features and retrain.")
    suspicious_features = suspicious['feature'].tolist()
else:
    print("   ✓ No suspicious features found (all correlations < 0.9)")
    suspicious_features = []

print("\n" + "="*70)
```

```python
print("Creating train-test split for binary classification...")

X_train_bin, X_test_bin, y_train_bin, y_test_bin = train_test_split(
    X, y_binary, test_size=0.2, random_state=42, stratify=y_binary
)

print("✓ Binary Classification Split:")
print(f"Training set: {X_train_bin.shape}")
print(f"Testing set: {X_test_bin.shape}")
print(f"\nTraining label distribution:")
print(pd.Series(y_train_bin).value_counts())
print(f"\nTesting label distribution:")
print(pd.Series(y_test_bin).value_counts())
```

```python
print("Creating train-test split for multi-class classification...")

X_train_multi, X_test_multi, y_train_multi, y_test_multi = train_test_split(
    X, y_multiclass, test_size=0.2, random_state=42, stratify=y_multiclass
)

print("✓ Multi-class Classification Split:")
print(f"Training set: {X_train_multi.shape}")
print(f"Testing set: {X_test_multi.shape}")
print(f"\nTraining attack distribution (top 10):")
print(pd.Series(y_train_multi).value_counts().head(10))
```

```python
print("Scaling features with StandardScaler...")

# Initialize scalers
scaler_bin = StandardScaler()
scaler_multi = StandardScaler()

# Scale binary classification features
X_train_bin_scaled = scaler_bin.fit_transform(X_train_bin)
X_test_bin_scaled = scaler_bin.transform(X_test_bin)

# Scale multi-class classification features
X_train_multi_scaled = scaler_multi.fit_transform(X_train_multi)
X_test_multi_scaled = scaler_multi.transform(X_test_multi)

print("✓ Feature scaling completed!")
```

```python
print("Training Random Forest for Binary Classification...")
print("(This may take several minutes for large datasets)")

rf_bin = RandomForestClassifier(
    n_estimators=100,
    max_depth=20,
    min_samples_split=100,  # Increased for faster training on large data
    min_samples_leaf=50,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1,
    verbose=1
)

rf_bin.fit(X_train_bin_scaled, y_train_bin)
print("\n✓ Random Forest (Binary) training completed!")
```

```python
print("Evaluating Random Forest (Binary)...")

# Predictions
y_pred_rf_bin = rf_bin.predict(X_test_bin_scaled)
y_pred_proba_rf_bin = rf_bin.predict_proba(X_test_bin_scaled)[:, 1]

# Metrics
acc_rf_bin = accuracy_score(y_test_bin, y_pred_rf_bin)
prec_rf_bin = precision_score(y_test_bin, y_pred_rf_bin)
rec_rf_bin = recall_score(y_test_bin, y_pred_rf_bin)
f1_rf_bin = f1_score(y_test_bin, y_pred_rf_bin)
roc_auc_rf_bin = roc_auc_score(y_test_bin, y_pred_proba_rf_bin)

print("\n" + "="*60)
print("RANDOM FOREST - BINARY CLASSIFICATION RESULTS")
print("="*60)
print(f"Accuracy:  {acc_rf_bin:.4f}")
print(f"Precision: {prec_rf_bin:.4f}")
print(f"Recall:    {rec_rf_bin:.4f}")
print(f"F1-Score:  {f1_rf_bin:.4f}")
print(f"ROC-AUC:   {roc_auc_rf_bin:.4f}")
print("="*60)

print("\nClassification Report:")
print(classification_report(y_test_bin, y_pred_rf_bin, 
                          target_names=['Attack', 'Benign']))
```
